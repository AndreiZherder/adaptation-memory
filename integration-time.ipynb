```python
from numpy import argmax
#v. 20251129.1
import matplotlib.pyplot as plt
import numpy as np


input_dir = "./data/integration_time/"
filename = "05_15.10.2025_c1 RED сone.txt"
fs = 500
prefix = 1
left = 0
right = 0.5
y = []
with open(input_dir + filename, 'r', encoding='utf-8') as file:
    for line in file:
        try:
            line = line.replace(',', '.')
            y.append(float('0' if line == '.' else line))
        except ValueError:
            continue
n = len(y)
y = np.array(y)
t = np.linspace(-prefix, (n - 1) / fs - prefix, n)
t = np.array(t)

l = round(prefix * fs + left * fs)
r = round(prefix * fs + right * fs)
maxi = argmax(y)
tmax = t[maxi]
ymax = y[maxi]
s = sum(y[i] for i in range(l, r + 1)) / ymax / fs
print(f'tmax = {tmax:.3f}, ymax = {ymax:.3f}, s = {s:.3f}')

fig, ax = plt.subplots()
ax.plot(t, y, 'b')
ax.axvline(x=left, color='r')
ax.axvline(x=right, color='r')
fig.set_figwidth(16)
fig.set_figheight(8)
ax.set_title('filename')
ax.set_xlabel('time, s')
ax.set_ylabel('response')
ax.grid()
```
